```python
import numpy as np
import numpy as np
import finufft
import matplotlib.pyplot as pl
import time
tic = time.time()
pi = np.pi

c = 3e8
G = 6.67e-11
const = 96/5*pi**(8/3)*(G/c**3)**(5/3)

f_max = 200
f_nufft = 3*f_max

def t_max_calc(f0, beta, f_max):
    temp0 = 0.375/beta
    temp1 = 1-(f0/f_max)**(8/3)
    return temp0*temp1

def nufft(beta, data, t, bin_no):
    bins = np.arange(bin_no) - bin_no//2
    
    tau = -3/5*(1-8/3*beta*t)**(5/8)/beta
    scale = (2*pi)/(tau[-1]-tau[0])
    tau *= scale
    start_diff = -pi-tau[0]
    tau += start_diff
    nufft_amp = finufft.nufft1d1(tau, data, bin_no)
    nufft_freq = -bins*scale/(2*pi)
    nufft_power = np.abs(nufft_amp/len(data))**2
    
    sort_idx = np.argsort(nufft_power)
    thresh_idx = np.searchsorted(nufft_power, np.array([0.01]), sorter=sort_idx)[0]
    return nufft_power[sort_idx[-1]]

def main_calc(f, M, T_obs=1e3):
    beta = const*f**(8/3)*M**(5/3)
    T_obs = min(1e3, t_max_calc(f, beta, 200))
    
    t_nufft = np.arange(0, T_obs, 1/f_nufft)
    phi = -6*pi/5*f*(1-8./3.*(beta)*t_nufft)**(5/8)/beta
    phi = np.mod(phi,2*pi)
    signal_nufft = 1*np.exp(1j*phi)
#    signal_nufft[:int(1e3)] = np.zeros((int(1e3)))
#    signal_nufft[-int(1e3):] = np.zeros((int(1e3)))
    
    ref_power = nufft(beta, signal_nufft, t_nufft, len(signal_nufft))
    result = np.array([nufft((1+i)*beta, signal_nufft, t_nufft, len(signal_nufft)) for i in offset_arr])  
#    mempool.free_all_blocks()
    
    return result
```

##  Coverage test - randomly chosen f0, Mc, piece size

```python
def coherent_analysis(data, tau, pieces, bin_no):
    """ 
    Make sure pieces is a factor of len(data)
    Default is interpolation
    """
    smaller_size = len(data)//pieces
    temp_data = data.reshape((pieces, smaller_size))
    temp_tau = tau.reshape((pieces, smaller_size))
    nufft_amps = np.empty((pieces, bin_no), dtype=complex)
    for i in range(pieces):
        nufft_amps[i] = finufft.nufft1d1(temp_tau[i], temp_data[i], bin_no)
    return np.sum(nufft_amps, axis=0)
```

```python
N = 10
pieces_arr = [2, 5, 10]
result_arr = []

for i in range(N):
    f0 = np.random.uniform(20, 30, 1)[0]
    Mc = 10**np.random.uniform(-1.3, -3, 1)[0] * 2e30
    pieces = 2 # pieces_arr[np.random.randint(0, 3, 1)[0]]
    beta = const*f0**(8/3)*Mc**(5/3)
    T_obs = min(1e4, t_max_calc(f0, beta, 200))

    t_nufft = np.arange(0, T_obs, 1/f_nufft)
    phi = -6*pi/5*f0*(1-8./3.*(beta)*t_nufft)**(5/8)/beta
    signal_nufft = 1*np.exp(1j*phi)

    bin_no = len(signal_nufft)
    data = signal_nufft
    t = t_nufft
    bins = np.arange(bin_no) - bin_no//2

    resampled_t = -3/5*(1-8/3*beta*t)**(5/8)/beta
    scale = (2*pi)/(resampled_t[-1]-resampled_t[0])
    tau = resampled_t*scale
    start_diff = -pi-tau[0]
    tau += start_diff
    nufft_amp = finufft.nufft1d1(tau, data, bin_no)
    nufft_freq = -bins*scale/(2*pi)
    nufft_power = np.abs(nufft_amp/len(data))**2
    coherent_amp = coherent_analysis(data, tau, pieces)
    result_arr.append(np.allclose(nufft_amp, coherent_amp))
```

```python
np.all(result_arr)
```

## Understanding what's happening

```python
%matplotlib inline

T_obs = 1e0
t_nufft = np.linspace(0, T_obs, 4096)
signal = np.sin(2*pi*1/T_obs*t_nufft)

bin_no = len(signal)
bins = np.arange(bin_no) - bin_no//2

nufft_amp = finufft.nufft1d1(t_nufft*2*pi, signal.astype(complex), bin_no)
nufft_power = np.abs(nufft_amp/len(data))**2

time1 = t_nufft[:len(t_nufft)//2]
time2 = t_nufft[len(t_nufft)//2:]
data1 = signal[:len(t_nufft)//2]
data2 = signal[len(t_nufft)//2:]

nufft_amp1 = finufft.nufft1d1(time1*2*pi, data1.astype(complex), bin_no//2)
nufft_amp2 = finufft.nufft1d1(time2*2*pi, data2.astype(complex), bin_no//2)
coherent_amp = nufft_amp1 + nufft_amp2
fft_amp = np.fft.fftshift(np.fft.fft(signal))
fft_freqs = np.fft.fftshift(np.fft.fftfreq(len(signal), 
                                           t_nufft[1] - t_nufft[0]))
```

```python
pl.plot(time1, data1)
pl.plot(time2, data2)
```

```python
pl.plot(fft_freqs, np.abs(fft_amp)**2, label="Standard fft")
pl.plot(bins, np.abs(nufft_amp)**2, label = "NUFFT on whole signal")
pl.plot(0.5*bins[::2], np.abs(coherent_amp)**2, label = "Coherently combined NUFFT")
pl.xlim(-3, 3)
print("All 3 overlap exactly")
```
