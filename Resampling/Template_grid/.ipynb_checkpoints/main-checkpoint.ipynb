```python
import numpy as np
import numpy as np
import finufft
import matplotlib.pyplot as pl
import time
tic = time.time()
pi = np.pi

c = 3e8
G = 6.67e-11
const = 96/5*pi**(8/3)*(G/c**3)**(5/3)

f_max = 200
f_nufft = 3*f_max

def t_max_calc(f0, beta, f_max):
    temp0 = 0.375/beta
    temp1 = 1-(f0/f_max)**(8/3)
    return temp0*temp1

def nufft(beta, data, t, bin_no):
    bins = np.arange(bin_no) - bin_no//2
    
    tau = -3/5*(1-8/3*beta*t)**(5/8)/beta
    scale = (2*pi)/(tau[-1]-tau[0])
    tau *= scale
    start_diff = -pi-tau[0]
    tau += start_diff
    nufft_amp = finufft.nufft1d1(tau, data, bin_no)
    nufft_freq = -bins*scale/(2*pi)
    nufft_power = np.abs(nufft_amp/len(data))**2
    
    sort_idx = np.argsort(nufft_power)
    thresh_idx = np.searchsorted(nufft_power, np.array([0.01]), sorter=sort_idx)[0]
    return nufft_power[sort_idx[-1]]

def main_calc(f, M, T_obs=1e3):
    beta = const*f**(8/3)*M**(5/3)
    T_obs = min(1e3, t_max_calc(f, beta, 200))
    
    t_nufft = np.arange(0, T_obs, 1/f_nufft)
    phi = -6*pi/5*f*(1-8./3.*(beta)*t_nufft)**(5/8)/beta
    phi = np.mod(phi,2*pi)
    signal_nufft = 1*np.exp(1j*phi)
#    signal_nufft[:int(1e3)] = np.zeros((int(1e3)))
#    signal_nufft[-int(1e3):] = np.zeros((int(1e3)))
    
    ref_power = nufft(beta, signal_nufft, t_nufft, len(signal_nufft))
    result = np.array([nufft((1+i)*beta, signal_nufft, t_nufft, len(signal_nufft)) for i in offset_arr])  
#    mempool.free_all_blocks()
    
    return result
```

## Entire curve

```python
f0 = 20
Mc = 1e-2 * 2e30
beta = const*f0**(8/3)*Mc**(5/3)
offset_arr = np.logspace(-7, -2, 50)

result = main_calc(f0, Mc)

np.savetxt('finufft_ratio_curve.txt', result)
pl.semilogx(offset_arr, result, 'o')
pl.title(r'$f_0 = 20, \, M_c = 10^{-2}M_\odot$')
pl.xlabel(r"$\Delta \beta / \beta$")
pl.ylabel("Recovered power")
# pl.savefig('finufft_power_ratio.png')

toc=time.time()
print((toc-tic)/60)
```

```python
beta = const*f0**(8/3)*Mc**(5/3)
T_obs = min(1e3, t_max_calc(f0, beta, 200))

t_nufft = np.arange(0, T_obs, 1/f_nufft)
phi = -6*pi/5*f0*(1-8./3.*(beta)*t_nufft)**(5/8)/beta
# phi = np.mod(phi,2*pi)
signal_nufft = 1*np.exp(1j*phi)

#nufft
bin_no = len(signal_nufft)
data = signal_nufft
t = t_nufft
bins = np.arange(bin_no) - bin_no//2

resampled_t = -3/5*(1-8/3*beta*t)**(5/8)/beta
scale = (2*pi)/(resampled_t[-1]-resampled_t[0])
tau = resampled_t*scale
start_diff = -pi-tau[0]
tau += start_diff
nufft_amp = finufft.nufft1d1(tau, data, bin_no)
nufft_freq = -bins*scale/(2*pi)
nufft_power = np.abs(nufft_amp/len(data))**2

sort_idx = np.argsort(nufft_power)
nufft_power[sort_idx[-1]]
```

Without offset is there any dephasing?

```python
np.allclose(2*pi*f0*(resampled_t - resampled_t[0]) + phi[0], phi)
# no
```

With offset is the dephasing correct? correctly predicted by the analytical estimation?

```python
def approx_dephasing(beta, T_obs, f0, delta_beta):
    temp0 = 6*f0*pi*(-1+beta*T_obs)*delta_beta
    temp1 = 5*beta**2*(1-8/3*beta*T_obs)**(3/8)
    const = 6*pi*f0*delta_beta/(5*beta**2)
    return -temp0/temp1 - const
```

```python
delta_beta = offset_arr[-10]*beta
bin_no = len(signal_nufft)
data = signal_nufft
t = t_nufft
bins = np.arange(bin_no) - bin_no//2

resampled_t = -3/5*(1-8/3*(beta+delta_beta)*t)**(5/8)/(beta+delta_beta)
scale = (2*pi)/(resampled_t[-1]-resampled_t[0])
tau = resampled_t*scale
start_diff = -pi-tau[0]
tau += start_diff
nufft_amp = finufft.nufft1d1(tau, data, bin_no)
nufft_freq = -bins*scale/(2*pi)
nufft_power = np.abs(nufft_amp/len(data))**2

sort_idx = np.argsort(nufft_power)
nufft_power[sort_idx[-1]]
```

```python
pl.plot(t, 2*pi*f0*(resampled_t - resampled_t[0]) + phi[0]-phi, label = "Numerical dephasing")
pl.plot(t, approx_dephasing(beta, t, f0, delta_beta), label = "Analytic dephasing")
pl.legend()
pl.ylabel(r"$\Delta \, \phi$ (radians)")
pl.xlabel("Time (s)")
pl.title("Checking numerical and analytic dephasing agree")
# Visually looks the overlapping, so analytic dephasing is correct
```

```python
# The approximation does break down a bit
pl.plot(t, 2*pi*f0*(resampled_t - resampled_t[0]) + phi[0]-phi - approx_dephasing(beta, t, f0, delta_beta))
pl.title("Difference between numerical and analytic dephasing")
pl.ylabel(r"$\Delta \, \phi$ (radians)")
pl.xlabel("Time (s)")
```

## Let's calculate the max step now

```python
#Analytic
def max_step(beta, T_obs, f0, phi_thresh):
    temp0 = -5*phi_thresh*beta**2/(6*pi*f0)
    temp1 = 1+(-1+beta*T_obs)/(1-8/3*beta*T_obs)**(3/8)
    return temp0/temp1
```

```python
#Check the inversion produces the identity
approx_dephasing(beta, T_obs, f0, max_step(beta, T_obs, f0, 1))
```

```python
beta_arr = np.logspace(-9,-4,100)
step_arr = [max_step(i, 1e3, f0, pi) for i in beta_arr]
pl.loglog(beta_arr, step_arr/beta_arr)
pl.title(r"Maximum step vs beta for T$_{obs}=1e3$")
pl.xlabel(r"$\beta$")
pl.ylabel(r"$\Delta \, \beta_{max}/\beta$")
```

## Numerically estimating step

```python
#Numerically inverting the dephasing equation
from scipy.optimize import fsolve
def temp(offset, f0=f0, beta=beta, t=T_obs, phi_thresh=pi/4):
    delta_beta = offset*beta
    temp0 = 6*f0*pi*(-1+beta*t)*delta_beta
    temp1 = 5*beta**2*(1-8/3*beta*t)**(3/8)
    const = 6*pi*f0*delta_beta/(5*beta**2)
    return -temp0/temp1 - const - phi_thresh

def numerical_step(f0=f0, beta=beta, t=T_obs, phi_thresh=pi/4):
    return fsolve(temp, 1e0, args=(f0, beta, T_obs, phi_thresh))
```

```python
#Check the inversion produces the identity
approx_dephasing(beta, T_obs, f0, beta*numerical_step(f0, beta, T_obs, 1))
```

```python
# The numerical and analytic estimations give the same results
beta_arr = np.logspace(-6, -9, 100)
numerical_result = [numerical_step(f0, i, T_obs, pi/4) for i in beta_arr]
step_arr = [max_step(i, T_obs, f0, pi/4) for i in beta_arr]

pl.loglog(beta_arr, [numerical_step(f0, i, T_obs, pi/4) for i in beta_arr], label='Numerical')
pl.loglog(beta_arr, step_arr/beta_arr, label = "Analytic")
pl.legend()
pl.title(r"Maximum step vs beta for T$_{obs}=1e3$")
pl.xlabel(r"$\beta$")
pl.ylabel(r"$\Delta \, \beta_{max}/\beta$")
```

## Testing the max steps lead to consistent power loss

```python
beta = 1e-4
T_obs = 1e3
step = max_step(beta, T_obs, f0, 2*pi)

t_nufft = np.arange(0, T_obs, 1/f_nufft)
phi = -6*pi/5*f0*(1-8./3.*(beta)*t_nufft)**(5/8)/beta
# phi = np.mod(phi,2*pi)
signal_nufft = 1*np.exp(1j*phi)

#nufft
ref_power = nufft(beta, signal_nufft, t_nufft, len(signal_nufft))
result = nufft(beta+step, signal_nufft, t_nufft, len(signal_nufft))
print(result/ref_power)
```

```python
beta_arr = np.logspace(-9, -4, 30)
result_arr = []
for i in beta_arr:
    beta = i
    T_obs = 1e3
    step = max_step(beta, T_obs, f0, pi)

    t_nufft = np.arange(0, T_obs, 1/f_nufft)
    phi = -6*pi/5*f0*(1-8./3.*(beta)*t_nufft)**(5/8)/beta
    # phi = np.mod(phi,2*pi)
    signal_nufft = 1*np.exp(1j*phi)

    #nufft
    ref_power = nufft(beta, signal_nufft, t_nufft, len(signal_nufft))
    result = nufft(beta+step, signal_nufft, t_nufft, len(signal_nufft))
    result_arr.append(result/ref_power)
```

```python
# Consistently recovers ~95% of max power for max dephasing of pi
# Some small variation but not huge
pl.semilogx(beta_arr, result_arr)
pl.title(r"Power recovered by taking $\Delta \beta_{max}$")
pl.xlabel(r"$\beta$")
pl.ylabel("Recovered power")
```

## Now letting T_obs = signal duration instead of fixed

```python
def step_vs_fMax(beta, f0, f_max, phi_thresh):
    T_obs = t_max_calc(f0, beta, f_max)
    temp0 = -5*phi_thresh*beta**2/(6*pi*f0)
    temp1 = 1+(-1+beta*T_obs)/(1-8/3*beta*T_obs)**(3/8)
    return temp0/temp1
```

```python
beta_arr = np.logspace(-9,-1,100)
step_arr = [step_vs_fMax(i, 20, 200, pi) for i in beta_arr]
step_arr1 = [step_vs_fMax(i, 150, 200, pi) for i in beta_arr]
pl.loglog(beta_arr, step_arr/beta_arr, label = 'f0 = 20Hz')
pl.loglog(beta_arr, step_arr1/beta_arr, label = 'f0 = 150Hz')
pl.grid()
pl.legend()
pl.title(r"Maximum step vs beta for T$_{obs}=$ signal duration")
pl.xlabel(r"$\beta$")
pl.ylabel(r"$\Delta \, \beta_{max}/\beta$")
```
