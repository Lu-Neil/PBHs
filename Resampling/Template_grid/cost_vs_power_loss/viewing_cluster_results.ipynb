```python
import numpy as np
import matplotlib.pyplot as pl
```

```python
result = np.loadtxt("./cluster_results/result_power_loss_scaling.txt")
```

```python
scaling_30 = result[:,0]
scaling_30 = np.reshape(scaling_30, (10, 3))
scaling_60 = result[:,1]
scaling_60 = np.reshape(scaling_60, (10, 3))
scaling_90 = result[:,2]
scaling_90 = np.reshape(scaling_90, (10, 3))
```

```python
pl.loglog(scaling_30[:,0], scaling_30[:,1], 'o', color='tab:blue', label="Recovered power = 30%")
pl.loglog(scaling_60[:,0], scaling_60[:,1], 'o', color='tab:orange', label="Recovered power = 60%")
pl.loglog(scaling_90[:,0], scaling_90[:,1], 'o', color='tab:green', label="Recovered power = 90%")
pl.legend()
```
