Makes nufft_estimator "close to" strobo_estimator and np.average(h0). No idea why this works, power spectrum is similar before the multiplication. Possibly because of different normalisation conventions

```python
import numpy as np
import matplotlib.pyplot as pl
from five_vec import five_vec
import finufft
```

```python
c = 3e8
G = 6.67e-11
pi = np.pi
const = 96/5*pi**(8/3)*(G/c**3)**(5/3)
```

```python
def t_max_calc(f0, beta, f_max):
    temp0 = 0.375/beta
    temp1 = 1-(f0/f_max)**(8/3)
    return temp0*temp1

def f_calc(f0, beta, t):
    return f0*(1-8/3*beta*t)**(-3/8)
```

```python
def nufft(beta, data, t, bin_no):
    bins = np.arange(bin_no) - bin_no//2

    tau = -3/5*(1-8/3*beta*t)**(5/8)/beta
    scale = (2*pi)/(tau[-1]-tau[0])
    tau *= scale
    start_diff = -pi-tau[0]
    tau += start_diff
    nufft_amp = finufft.nufft1d1(
        tau.astype(np.float64), data.astype(complex), bin_no, eps=1e-12, upsampfac=2.0)/len(data)
    nufft_freq = -bins*scale/(2*pi)
    nufft_power = np.abs(nufft_amp)**2
    return nufft_freq, nufft_amp
```

## Comparison to strobo

```python
def strobo(beta, data, t, f_ratio, f_signal):
    new_t = -3/5*(1-8/3*beta*t)**(5/8)/beta
    f_new = f_signal/f_ratio
    
    new_t *= f_new #changes the frequency to the new downsampled frequency (approximately, not exactly)
    floor_t = np.floor(new_t) #For some reason floor works better than round. Take it to the nearest time index which are integers
    idx = np.nonzero(np.diff(floor_t)) #The step that downsamples
    resampled = data[idx]
    t_out = (new_t[idx])/f_new
    return (resampled, t_out)
```

```python
# Signal parameters
f0 = 20.5
Mc = 5e-2 * 2e30
beta = const*f0**(8/3)*Mc**(5/3)

f_max = 200
n_days = 3
side_day = 1e4/n_days #86164.09053083288
T_obs = n_days*side_day

f_strobo = 32*f_max
nt = round(f_strobo*T_obs)
t_strobo = (np.arange(nt)/f_strobo).astype(np.longdouble)
d = 8e3*3.086e+16 # 8kpc

# Amp modulation
params = dict(ra = 0.5, dec = 0.5, eta = 0.5, psi = 0.5, lat = 0.5, 
              lng = 0.5, az = 0.5, side_day=side_day)
sidereal = five_vec(**params)
sidereal.compute_H()
sidereal.compute_A(t_strobo)
sidereal.compute_5vec()
amp_modulation = sidereal.amp_modulation
```

```python
# Signal creation
phi = -6*pi/5*f0*(1-8./3.*(beta)*t_strobo)**(5/8)/beta
freq_signal = 1*np.exp(1j*phi)

# amp modulation
f = f_calc(f0, beta, t_strobo)
h0 = 4/d*(G*Mc/c**2)**(5/3)*(pi*f/c)**(2/3)
signal_strobo = h0*amp_modulation*freq_signal
```

```python
# Strobo
strobo_data, strobo_time = strobo(beta, signal_strobo, t_strobo, f_ratio = 8, f_signal=f_strobo)
strobo_amp = np.fft.fftshift(np.fft.fft(strobo_data))/len(strobo_data)
strobo_freq = np.fft.fftshift(np.fft.fftfreq(len(strobo_data), np.diff(strobo_time)[0]))

f0_idx = abs(strobo_freq-f0).argmin()
strobo_X = np.empty((5), dtype=complex)
strobo_X[0] = strobo_amp[f0_idx-n_days*2]
strobo_X[1] = strobo_amp[f0_idx-n_days*1]
strobo_X[2] = strobo_amp[f0_idx]
strobo_X[3] = strobo_amp[f0_idx+n_days*1]
strobo_X[4] = strobo_amp[f0_idx+n_days*2]

strobo_estimator = abs(np.dot(strobo_X, np.conj(sidereal.A))/np.sum(np.abs(sidereal.A)**2))
```

```python
nufft_downsamp = 8
signal_nufft = signal_strobo[::nufft_downsamp]
t_nufft = t_strobo[::nufft_downsamp]

params = dict(ra = 0.5, dec = 0.5, eta = 0.5, psi = 0.5, lat = 0.5, 
              lng = 0.5, az = 0.5, side_day=side_day)
sidereal = five_vec(**params)
sidereal.compute_H()
sidereal.compute_A(t_nufft)
sidereal.compute_5vec()
amp_modulation = sidereal.amp_modulation
```

```python
1/np.sqrt(2*np.pi)
```

```python
nufft_freq, nufft_amp = nufft(beta, signal_nufft, t_nufft, len(signal_nufft))
nufft_idx = abs(nufft_freq-f0).argmin()
nufft_X = np.empty((5), dtype=complex)
nufft_X[0] = nufft_amp[nufft_idx+n_days*2]
nufft_X[1] = nufft_amp[nufft_idx+n_days*1]
nufft_X[2] = nufft_amp[nufft_idx]
nufft_X[3] = nufft_amp[nufft_idx-n_days*1]
nufft_X[4] = nufft_amp[nufft_idx-n_days*2]

nufft_estimator = abs(np.dot(nufft_X, np.conj(sidereal.A))/np.sum(np.abs(sidereal.A)**2))
```

```python
print(strobo_estimator/np.average(h0))
print(2*nufft_estimator/np.average(h0))
```
