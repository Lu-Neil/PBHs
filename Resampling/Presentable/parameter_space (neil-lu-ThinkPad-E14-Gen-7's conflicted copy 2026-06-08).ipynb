## Parameter space

```python
def beta_calc(f, M):
    M_temp = M*2e30
    return const*f**(8/3)*M_temp**(5/3)

def beta_approx(f, M):
    return 3.7e-4*(M/1e-2)**(5/3)*(f/200)**(8/3)
```

```python
fspace = np.linspace(40, 60, 1000)
Mspace = np.logspace(-4, -1, 1000)

[fgrid, Mgrid] = np.meshgrid(fspace, Mspace)
betaGrid = beta_calc(fgrid, Mgrid)
betaApprox = beta_approx(fgrid, Mgrid)
np.allclose(betaApprox, betaGrid)
```

```python
pl.contourf(fspace, np.log10(Mspace), np.log10(betaGrid))
pl.colorbar()
pl.title(r'$log(\beta)$')
pl.xlabel("Frequency (Hz)")
pl.ylabel(r'log$_{10}(M_c/M_\odot$)')
```

```python
def t_max_calc(f0, M, f_max):
    temp0 = (M/1e-2)**(-5/3)
    temp1 = (f0/200)**(-8/3) - (f_max/200)**(-8/3)
    return 1013.75*temp0*temp1

def t_max_calc1(f0, beta, f_max):
    temp0 = 0.375/beta
    temp1 = 1-(f0/f_max)**(8/3)
    return temp0*temp1
```

```python
np.allclose(t_max_calc(fgrid, Mgrid, 200), 
                   t_max_calc1(fgrid, betaGrid, 200))
```

```python
tMax_grid = t_max_calc(fgrid, Mgrid, fspace[-1])
# tMax_grid[tMax_grid>3e7] = 3e7
```

```python
SECONDS_IN_HOUR = 60*60
pl.contourf(fspace, np.log10(Mspace), np.log10(tMax_grid / SECONDS_IN_HOUR))
cbar = pl.colorbar()
pl.contour(fspace, np.log10(Mspace), np.log10(tMax_grid / SECONDS_IN_HOUR), [np.log10(86400 / SECONDS_IN_HOUR), np.log10(30*86400 / SECONDS_IN_HOUR)], colors = ['darkorange', 'red'], linestyles = '--')
# pl.contour(fspace, np.log10(Mspace), np.log10(tMax_grid), [], colors = 'red', linestyles = '--')
cbar.ax.set_ylabel(r'$\log_{10}$(signal duration) [hours]')
pl.xlabel("Frequency (Hz)")
pl.ylabel(r'log$_{10}(M_c/M_\odot$)')

from matplotlib.lines import Line2D
line0 = Line2D([0], [0], label='1 day', color='darkorange', ls = '--')
line1 = Line2D([0], [0], label='1 month', color='r', ls = '--')
pl.legend(handles = [line0, line1])
# pl.savefig("Signal_duration", format = "jpeg")
```

```python
def f_calc(f0, t, beta):
    return f0*np.power(1-8/3*beta*t,-3/8)
```

```python
fMax_grid = f_calc(fgrid, tMax_grid, betaGrid)
```

```python
pl.contourf(fspace, np.log10(Mspace), fMax_grid)
pl.colorbar()
pl.title(r'$f(t_{max})$ [Hz]')
pl.xlabel("Frequency (Hz)")
pl.ylabel(r'log$_{10}(M_c/M_\odot$)')
```

```python
pl.contourf(fspace, np.log10(Mspace), np.log10(tMax_grid*betaGrid))
pl.colorbar()
pl.title(r'$\beta t_{max}$')
pl.xlabel("Frequency (Hz)")
pl.ylabel(r'log$_{10}(M_c/M_\odot$)')
```
