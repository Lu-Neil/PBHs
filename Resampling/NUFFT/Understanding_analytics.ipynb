```python
import numpy as np
import matplotlib.pyplot as pl
import numpy.fft as fft
pi = np.pi
```

```python
w = 10
t = np.linspace(-pi, pi, 2**12)
d = np.sin(w*t)
```

```python
pl.plot(t, d)
```

```python
amps = fft.fftshift(fft.fft(d))
power = np.abs(amps)**2
freqs = fft.fftshift(fft.fftfreq(len(d), t[1]-t[0]))
```

```python
pl.plot(freqs, power, 'o')
pl.axvline(w/(2*pi), c='r', ls='--')
pl.axvline(-w/(2*pi), c='r', ls='--')
pl.xlim(-5, 5)
```

```python
t_padded = np.linspace(-pi, pi, 2*len(t))
d_padded = np.zeros(2*len(t))
d_padded[::2] = d
```

```python
pl.plot(t_padded, d_padded, 'o')
```

```python
amps_padded = fft.fftshift(fft.fft(d_padded))
power_padded = np.abs(amps_padded)**2
freqs_padded = fft.fftshift(fft.fftfreq(len(d_padded), t[1]-t[0]))
```

```python
pl.plot(freqs_padded, power_padded, 'o')
pl.axvline(w/(2*pi), c='r', ls='--')
pl.axvline(-w/(2*pi), c='r', ls='--')
pl.xlim(325, 326)
```
