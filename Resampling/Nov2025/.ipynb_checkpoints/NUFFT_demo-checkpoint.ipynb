```python
import numpy as np
import finufft
import matplotlib.pyplot as pl
from resampler import Resampler

pi = np.pi
```

## Monochromatic signal

```python
t = np.linspace(4*pi,6*pi, int(2**15))
f0 = 8 # rad/s
phi = f0*t
signal = np.real(1*np.exp(-1j*phi))
```

```python
resampler = Resampler()
resampler.timeseries = signal
resampler.resampled_time = t
resampler.nufft()
pl.plot(resampler.freqs, resampler.power_normalized, 'o')
pl.xlim(f0 - 10, f0 + 10)
```

```python
resampler.nufft_real()
pl.plot(resampler.freqs, resampler.power_normalized, 'o')
pl.xlim(f0 - 10, f0 + 10)
```

## With $\dot{f}$

```python
t = np.linspace(0*pi,16*pi, 2**14)
f0 = 2 # rad/s
fdot = 0.1 # rad/s^2 
phi = f0*t+fdot*t**2
signal = np.real(1*np.exp(-1j*phi))
tau = t*(1+fdot/f0*t)
```

```python
resampler = Resampler()
resampler.timeseries = signal
resampler.resampled_time = tau
resampler.nufft_real()
pl.plot(resampler.freqs, resampler.power_normalized, 'o')
pl.xlim(f0 - 1, f0 + 1)
```

## beta

```python
c = 3e8
G = 6.67e-11
pi = np.pi
const = 96/5*pi**(8/3)*(G/c**3)**(5/3)

f0 = 2**3
Mc = 3e-2* 2e30
f_max = 2**6
T_obs = 2**10
beta = const*f0**(8/3)*Mc**(5/3)
f_signal = 4*f_max
nt = round(f_signal*T_obs)
t = (np.arange(nt)/f_signal)

phi = 6*np.pi/5*f0*(1-8./3.*(beta)*t)**(5/8)/beta
tau = 6*np.pi/5*(1-8/3*beta*t)**(5/8)/beta
signal = np.real(1*np.exp(-1j*phi))
```

```python
resampler = Resampler()
resampler.timeseries = signal
resampler.resampled_time = tau
resampler.nufft_real()
pl.plot(resampler.freqs, resampler.power_normalized, 'o')
pl.xlim(f0 - 0.05, f0 + 0.05)
```
